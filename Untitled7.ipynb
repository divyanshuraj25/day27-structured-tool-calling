```python
from pydantic import BaseModel, Field
from google.genai import types

# 1. Output ka Structured Schema define karo
class MovieInfo(BaseModel):
    title: str = Field(description="Movie ka naam")
    year: int = Field(description="Release year")
    genre: list[str] = Field(description="Movie ke genres")
    summary: str = Field(description="Chhota sa hindi/english plot summary")

# 2. Function Calling ke liye Tool Function define karo
def get_current_weather(location: str) -> str:
    """Kisi city ka current weather return karta hai."""
    # Dummy tool response for testing
    return f"{location} me abhi mausam saaf hai aur temperature 28°C hai."

print("Schema aur Tools successfully define ho gaye!")
```

```python
# A. Structured JSON Output Test (Pydantic Schema ke sath)
print("--- 1. Structured Output Test ---")
structured_response = client.models.generate_content(
    model="gemini-3.8-flash",
    contents="Inception movie ki details provide karo.",
    config=types.GenerateContentConfig(
        response_mime_type="application/json",
        response_schema=MovieInfo,
    ),
)
print("Structured Output:\n", structured_response.text)

# B. Function Calling / Tool Test
print("\n--- 2. Function Calling Test ---")
tool_response = client.models.generate_content(
    model="gemini-3.8-flash",
    contents="Mumbai me mausam kaisa hai?",
    config=types.GenerateContentConfig(
        tools=[get_current_weather],
    ),
)

# Tool call check karo
if tool_response.function_calls:
    for call in tool_response.function_calls:
        print(f"Model ne Tool Call kiya: {call.name}")
        print(f"Arguments: {call.args}")
else:
    print("Normal Response:", tool_response.text)
```

```python
# 1. Model se Tool Call request lena
prompt = "Mumbai me mausam kaisa hai?"
chat_response = client.models.generate_content(
    model="gemini-3.8-flash",
    contents=prompt,
    config=types.GenerateContentConfig(
        tools=[get_current_weather],
    ),
)

# 2. Tool execute karna aur response bhejna
if chat_response.function_calls:
    for call in chat_response.function_calls:
        print(f"Function called: {call.name} with args {call.args}")

        # Actual Python function ko run kiya
        if call.name == "get_current_weather":
            tool_result = get_current_weather(**call.args)

            # Model ko tool ka output provide kiya
            final_response = client.models.generate_content(
                model="gemini-3.8-flash",
                contents=[
                    prompt,
                    chat_response.candidates[0].content,
                    types.Part.from_function_response(
                        name=call.name,
                        response={"result": tool_result}
                    )
                ]
            )
            print("\nFinal Model Answer:")
            print(final_response.text)
```

```python
import nest_asyncio
import uvicorn
from fastapi import FastAPI
from pydantic import BaseModel
import threading

# Colab ke event loop ko patch karo
nest_asyncio.apply()

app = FastAPI(title="Gemini Agent API")

class QueryRequest(BaseModel):
    query: str

@app.post("/chat")
def chat_endpoint(req: QueryRequest):
    response = client.models.generate_content(
        model="gemini-3.8-flash",
        contents=req.query,
        config=types.GenerateContentConfig(
            tools=[get_current_weather],
        ),
    )

    if response.function_calls:
        call = response.function_calls[0]
        if call.name == "get_current_weather":
            res = get_current_weather(**call.args)
            final_res = client.models.generate_content(
                model="gemini-3.8-flash",
                contents=[
                    req.query,
                    response.candidates[0].content,
                    types.Part.from_function_response(
                        name=call.name,
                        response={"result": res}
                    )
                ]
            )
            return {"type": "tool_executed", "function": call.name, "answer": final_res.text}

    return {"type": "direct_answer", "answer": response.text}

# Server ko background thread me start karo
def run_server():
    uvicorn.run(app, host="127.0.0.1", port=8000, log_level="warning")

server_thread = threading.Thread(target=run_server, daemon=True)
server_thread.start()
print("FastAPI server 127.0.0.1:8000 par start ho gaya hai!")
```

```python
# Updated Endpoint with Forced / Strict Tool Calling
@app.post("/chat")
def chat_endpoint(req: QueryRequest):
    # Model ko batate hain ki agar suitable tool hai toh call zarur kare
    response = client.models.generate_content(
        model="gemini-3.8-flash",
        contents=req.query,
        config=types.GenerateContentConfig(
            tools=[get_current_weather],
            tool_config=types.ToolConfig(
                function_calling_config=types.FunctionCallingConfig(
                    mode="ANY"  # Force karega model ko tool select karne ke liye
                )
            ),
        ),
    )

    if response.function_calls:
        call = response.function_calls[0]
        print(f"Triggered Tool: {call.name} with args: {call.args}")

        if call.name == "get_current_weather":
            # Actual python function call
            res = get_current_weather(**call.args)

            # Function response model ko wapas bhejte hain
            final_res = client.models.generate_content(
                model="gemini-3.8-flash",
                contents=[
                    req.query,
                    response.candidates[0].content,
                    types.Part.from_function_response(
                        name=call.name,
                        response={"result": res}
                    )
                ]
            )
            return {
                "type": "tool_executed",
                "function": call.name,
                "arguments": call.args,
                "answer": final_res.text
            }

    return {"type": "direct_answer", "answer": response.text}

print("Endpoint updated with strict tool calling!")
```
